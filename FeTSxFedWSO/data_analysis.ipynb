# Non-IID

With partitioning_1 method from FeTS, clients are not IID. Since we want to assess the effect of FedWSO, we don't need the pure performance. We can merge the small institutions to create 4 larger clients.

The clients can be composed as following:
- Client 1 -> Institution 1
- Client 2 -> Institution 18
- Client 3 -> Institutions: 4, 21, 13, 6, 20, 16
- Client 4 -> Institutions: 5, 3, 11, 15, 7, 12, 17, 10, 8, 22, 14, 2, 23, 19, 9

### Non-IID client compositions

```python
client_1_df_train = df[df["Partition_ID"] == 1]
client_1_df_train.to_csv("./client1_noniid_dataset.csv", index=False)
```

```python
client_2_df_train = df[df["Partition_ID"] == 18]
client_2_df_train.to_csv("./client2_noniid_dataset.csv", index=False)
```

```python
client_3_df_train = df[df["Partition_ID"].isin([4,21, 13, 6,])]
client_3_df_train.to_csv("./client3_noniid_dataset.csv", index=False)
```

```python
client_4_df_train = df[df["Partition_ID"].isin([5, 3, 11, 15, 7, 12, 17, 10, 8, 22, 14, 2, 23, 19,])]
client_4_df_train.to_csv("./client4_noniid_dataset.csv", index=False)
```

# IID

We can use the same splitting, but forcing the larger client to a fixed number samples. 

```python
import pandas as pd

# Load and save Non-IID client data
client_1_df_noniid = pd.read_csv("client1_noniid_dataset.csv")
client_1_df_noniid.to_csv("client1_noniid_dataset.csv", index=False)

client_2_df_noniid = pd.read_csv("client2_noniid_dataset.csv")
client_2_df_noniid.to_csv("client2_noniid_dataset.csv", index=False)

client_3_df_noniid = pd.read_csv("client3_noniid_dataset.csv")
client_3_df_noniid.to_csv("client3_noniid_dataset.csv", index=False)

client_4_df_noniid = pd.read_csv("client4_noniid_dataset.csv")
client_4_df_noniid.to_csv("client4_noniid_dataset.csv", index=False)
# Group into a list for easy handling
clients_noniid = [
    client_1_df_noniid,
    client_2_df_noniid,
    client_3_df_noniid,
    client_4_df_noniid
]
# Print Non-IID sample sizes
print("Non-IID client sample sizes and saved files:")
for idx, client_df in enumerate(clients_noniid):
    print(f"Client {idx+1}: {len(client_df)} samples → Saved as 'client{idx+1}_noniid_dataset.csv'")
```

Since our aim is to analyse the impact of the FedWSO on the result, we don't need to keep every case of the original FeTS dataset and we can limit each client to 100 samples.

```python
n_max_samples = 100
clients_iid = []

print("\nCreating IID client datasets with up to 100 samples each:")
for idx, client_df in enumerate(clients_noniid):
    limited_df = client_df.head(n_max_samples)
    clients_iid.append(limited_df)

    iid_path = f"client{idx+1}_iid_dataset.csv"
    limited_df.to_csv(iid_path, index=False)
    
    print(f"Client {idx+1}: {len(limited_df)} samples saved to {iid_path}")
```
